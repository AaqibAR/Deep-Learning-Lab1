```python
plt.hist(np.random.rand(100000), density=True, bins=100, histtype="step", color="blue", label="rand")

plt.hist(np.random.exponential(scale=1.0, size=100000), density=True, bins=100, histtype="step", color="red", label="exponential")

plt.hist(np.random.normal(0, 1, 100000), density=True, bins=100, histtype="step", color="green", label="normal")

plt.axis([-2.5, 2.5, 0, 1.1])

plt.legend(loc = "upper left")

plt.title("Random distributions")

plt.xlabel("Value")

plt.ylabel("Density")

plt.show()
```

```python
# Generate the three distributions
exp_data = np.random.exponential(scale=1.0, size=100000)
uniform_data = np.random.uniform(-1, 1, 100000)  # roughly matching the [-2.5, 2.5] view
normal_data = np.random.normal(0, 1, 100000)

plt.figure(figsize=(8, 5))

plt.hist(uniform_data, density=True, bins=100, histtype="step", color="blue", label="uniform")
plt.hist(normal_data, density=True, bins=100, histtype="step", color="green", label="normal")
plt.hist(exp_data, density=True, bins=100, histtype="step", color="red", label="exponential")

plt.axis([-2.5, 5, 0, 1.1])
plt.legend(loc="upper left")
plt.title("Random distributions")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D  # needed for 3D plotting

# Step 1: create x and y ranges
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

# Step 2: build the grid of all (x, y) combinations
X, Y = np.meshgrid(x, y)

# Step 3: compute Z at every grid point
Z = X**2 + Y**2

# Step 4: plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Z = X² + Y²')

plt.show()
```

```python
import pandas as pd

!wget -q https://elitedatascience.com/wp-content/uploads/2022/07/Pokemon.csv
df = pd.read_csv('Pokemon.csv', index_col=0, encoding='latin')
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

cols = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']
stats_df = df[cols]

pearson_corr = stats_df.corr(method='pearson')
spearman_corr = stats_df.corr(method='spearman')

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.heatmap(pearson_corr, annot=True, cmap='coolwarm', fmt='.2f', ax=axes[0])
axes[0].set_title('Pearson Correlation')

sns.heatmap(spearman_corr, annot=True, cmap='coolwarm', fmt='.2f', ax=axes[1])
axes[1].set_title('Spearman Correlation')

plt.tight_layout()
plt.show()
```
